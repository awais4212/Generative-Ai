### Text to Text Generation using Groq

```python
import os
from dotenv import load_dotenv
from groq import Groq

load_dotenv(".env.local")   # VS Code runs notebooks from their own folder, so this path works
GROQ_API_KEY = os.getenv("GROQ_API_KEY")

if not GROQ_API_KEY:
    raise ValueError("GROQ_API_KEY not found in .env.local")

client = Groq(api_key=GROQ_API_KEY)
```

```python
# Install first if needed:
# !pip install groq

# ---------------------------------------------
# STEP 2: FIND AN AVAILABLE TEXT MODEL
# ---------------------------------------------

def get_available_model():

    # Get models available to YOUR Groq account
    models = client.models.list()

    available_models = [model.id for model in models.data]

    print("Available Models:")
    for model in available_models:
        print("-", model)

    # Models we prefer, in order
    preferred_models = [
        "llama-3.3-70b-versatile",
        "openai/gpt-oss-20b",
        "openai/gpt-oss-120b",
        "llama-3.1-8b-instant"
    ]

    # Select the first available model
    for model in preferred_models:
        if model in available_models:
            print("\nSelected Model:", model)
            return model

    return None


# ---------------------------------------------
# STEP 3: TEXT-TO-TEXT FUNCTION
# ---------------------------------------------

def generate_text(prompt, model_name):

    response = client.chat.completions.create(

        model=model_name,

        messages=[
            {
                "role": "system",
                "content": """
You are a helpful AI assistant.
Answer clearly, naturally, and concisely.
"""
            },
            {
                "role": "user",
                "content": prompt
            }
        ],

        temperature=0.7,
        max_tokens=300
    )

    return response.choices[0].message.content


# ---------------------------------------------
# STEP 4: SELECT MODEL
# ---------------------------------------------

selected_model = get_available_model()


# ---------------------------------------------
# STEP 5: GENERATE TEXT
# ---------------------------------------------

if selected_model is not None:

    prompt = "How to Play Cricket"

    try:

        result = generate_text(
            prompt,
            selected_model
        )

        print("\n-----------------------------")
        print("USER:")
        print(prompt)

        print("\n-----------------------------")
        print("AI RESPONSE:")
        print(result)

    except Exception as error:

        print("\nAn error occurred:")
        print(error)

else:

    print("\nNo supported text-generation model was found.")
    print("Check your Groq account and model access.")
```

### Text to Speech Generation using gTTS

```python
from gtts import gTTS
from IPython.display import Audio, display


def text_to_speech(text, lang='en'):

    # Convert generated text into speech
    tts = gTTS(
        text=text,
        lang=lang,
        slow=False
    )

    # Save audio file
    audio_file = "generated_speech.mp3"
    tts.save(audio_file)

    # Return audio
    return Audio(audio_file, autoplay=True)


# --------------------------------------
# Use the Groq-generated result
# --------------------------------------

print("Converting generated text to speech...")

audio_output = text_to_speech(result)

display(audio_output)

print("Speech generated and played successfully!")
```
